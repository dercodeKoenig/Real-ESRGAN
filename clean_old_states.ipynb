```python
import os


for d in os.listdir("experiments"):
    d = "experiments/"+d
    print(d)
    try:
        states = os.listdir(d+"/training_states")
        l = []
        for i in states:
            if i.endswith(".state"):
                l.append(int(i.split(".")[0]))
        
        last_num = str(sorted(l)[-1])
        latest_name = last_num+".state"
        print(latest_name)
        
        for i in states:
            if i != latest_name:
                os.system("rm "+d+"/training_states/"+i)
                #print("rm "+d+"/training_states/"+i)
        
            
            
        models = os.listdir(d+"/models")
        for i in models:
            if not last_num in i:
                os.system("rm "+d+"/models/"+i)
    except Exception as e:
        print(e)
```
