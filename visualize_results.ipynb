```python
import os
from PIL import Image
import time
import IPython

path = "experiments/"+experiment+"/visualization/"
output_paths = {}
for d in os.listdir(path):
    try:
        hn = 0
        name = ""
        for i in os.listdir(os.path.join(path, d)):
            try:
                n = int(i.split(".")[0].split("_")[1])
                if n > hn:
                    hn = n
                    name = i
            except:
                pass
        print(name)
        output_path = os.path.join(path, d, name)
        Image.open(output_path).show()
        output_paths[d] = output_path
    except:pass
```

```python
from PIL import Image

def crop_image(img, x, y, w, h):
    # Define crop box: (left, upper, right, lower)
    crop_box = (x, y, x + w, y + h)
    return img.crop(crop_box)

def combine_images_horizontally(images, border_color=(0, 0, 0)):
    if not images:
        raise ValueError("Images list cannot be empty")
    
    # Get total width (including borders) and max height
    total_width = sum(img.width for img in images) + (len(images) - 1)  # Add 1px for each border
    max_height = max(img.height for img in images)
    
    # Create a new blank canvas with border color
    combined = Image.new("RGB", (total_width, max_height), border_color)
    
    # Paste images side by side with 1px spacing
    x_offset = 0
    for i, img in enumerate(images):
        combined.paste(img, (x_offset, 0))
        x_offset += img.width + 1  # Add 1px border after each image (except the last)
    
    return combined
```

```python
import cv2
import os
import re
from tqdm import tqdm
from IPython.display import Video, display

#for d in ["baboon", "comic", "photo1", "photo2"]:
for d in ["baboon", "comic", "photo1"]:
#for d in ["baboon", "comic"]:
        image_folder = "experiments/"+experiment+"/visualization/"+d
        
        # Regex pattern to extract number from filenames like [xyz]_123.png
        pattern = re.compile(r".*_(\d+)\.png$")
        
        # Find and sort image files by the number
        image_files = []
        for filename in os.listdir(image_folder):
            match = pattern.match(filename)
            if match:# and "000.png" in filename:
                number = int(match.group(1))
                image_files.append((number, os.path.join(image_folder, filename)))
        
        # Sort by extracted number
        image_files.sort(key=lambda x: x[0])
        
        # Ensure we have at least one image
        if not image_files:
            raise ValueError("No matching images found.")
        
        # Read first image to get frame size
        first_img = cv2.imread(image_files[0][1])
        height, width, layers = first_img.shape
        fps = 40  # Frames per second
        
        # Define the codec and create VideoWriter object
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # or 'XVID'
        video_path = "/".join([image_folder, "output_video.mp4"])
        out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))
        
        # Write each image as a frame to the video
        for _, file_path in tqdm(image_files, desc="Writing video"):
            img = cv2.imread(file_path)
            if img is not None:
                out.write(img)
        
        # Release everything if job is finished
        out.release()
        #print(f"Video saved as {video_path}")
        
        video_path2 = video_path+"_conv.mp4"
        !ffmpeg -y -i $video_path -vcodec libx264 -acodec aac -strict -2 $video_path2 > /dev/null 2>&1
        # Display video
        display(Video(video_path2, embed=True))
        
```
