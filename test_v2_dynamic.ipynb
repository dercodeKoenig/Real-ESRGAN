```python
import cv2
from PIL import Image

path = "tests/data/lq_4/comic.png"
scale = 6
img = cv2.imread(path, cv2.IMREAD_UNCHANGED)
print(img.shape)
Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
img = cv2.resize(
                img, (
                    int(img.shape[1] * scale),
                    int(img.shape[0] * scale),
                ), interpolation=cv2.INTER_CUBIC)
print(img.shape)
Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
from realesrgan import RealESRGANerV2
from realesrgan.archs.rrdb_pixelshuffle_v3_arch import RRDBNet_v3
from realesrgan.archs.rrdb_unet_arch import RRDB_UNet

#model = RRDBNet_v3(num_in_ch=3, num_out_ch=3, highway_channels=192, processing_channels=64, num_block=16, num_grow_ch=32, num_pre_upscale_ch=256, scale=1, use_attention=True)
#model_path = 'experiments/train_v3_dynamic_s/models/net_g_160000.pth'

#model = RRDBNet_v3(num_in_ch=3, num_out_ch=3, highway_channels=32, processing_channels=16, num_block=8, num_grow_ch=16, num_pre_upscale_ch=128, scale=1, use_attention=True)
#model_path = 'experiments/train_v3_dynamic_iterative_s/models/net_g_790000.pth'

model = RRDB_UNet(
    num_in_ch=3,
    num_out_ch=3,
    highway_channels_base=24,
    processing_channels_base=12,
    num_grow_ch_base = 6,
    ae_rrdb_blocks=3,
    ae_channel_multipliers = [1,2,4,12,36],
    use_attention=True,
    body_rrdb_blocks = 12
)
model_path = 'experiments/train_unet_dynamic_iterative_m/models/net_g_138472.pth'

upscaler = RealESRGANerV2(
    scale=1,
    model_path=model_path,
    model=model,
)
```

```python
out, _ = upscaler.enhance(img)
print(out.shape)
Image.fromarray(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
```
