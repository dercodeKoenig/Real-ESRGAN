```python
import cv2
import matplotlib.pyplot as plt
path = "tests/data/lq_4/comic.png"
scale = 8
img = cv2.imread(path, cv2.IMREAD_UNCHANGED)
print(img.shape)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
img = cv2.resize(
                img, (
                    int(img.shape[1] * scale),
                    int(img.shape[0] * scale),
                ), interpolation=cv2.INTER_CUBIC)
print(img.shape)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
from realesrgan import RealESRGANerV2
from realesrgan.archs.rrdb_pixelshuffle_v3_arch import RRDBNet_v3

model = RRDBNet_v3(num_in_ch=3, num_out_ch=3, highway_channels=192, processing_channels=64, num_block=16,
                   num_grow_ch=32, num_pre_upscale_ch=256, scale=1, use_attention=True)
model_path = 'experiments/train_v3_dynamic_s/models/net_g_160000.pth'
upscaler = RealESRGANerV2(
    scale=1,
    model_path=model_path,
    model=model,
)
```

```python
out, _ = upscaler.enhance(img)
print(out.shape)
plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
```

```python
out, _ = upscaler.enhance(out)
print(out.shape)
plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
```

```python
out, _ = upscaler.enhance(img)
print(out.shape)
plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))

img = img * 0.9 + out * 0.1
```
