```python
from basicsr.metrics import calculate_psnr, calculate_ssim
import cv2
import os
import numpy as np
from realesrgan.utils import RealESRGANerV2
from realesrgan.archs.rrdb_pixelshuffle_v3_arch import RRDBNet_v3

model = RRDBNet_v3(num_in_ch=3, num_out_ch=3, highway_channels=192, processing_channels=64, num_block=32,
                       num_grow_ch=32, num_pre_upscale_ch=256, scale=4, use_attention=True)

model_path = "../../../experiments/ft_lowgan_v3_attn_x4_m/models/net_g_35000.pth"
upscaler = RealESRGANerV2(
        scale=4,
        model_path=model_path,
        model=model,
)

lr_dir = "bicubic_4x/train/LR"
gt_dir = "bicubic_4x/train/HR"

psnr_list = []
ssim_list = []

for fname in os.listdir(lr_dir):
    lr_img = cv2.imread(os.path.join(lr_dir, fname), cv2.IMREAD_COLOR)
    gt_img = cv2.imread(os.path.join(gt_dir, fname), cv2.IMREAD_COLOR)
    print(lr_img.shape, gt_img.shape)

    sr_img, mode = upscaler.enhance(lr_img)

    sr_img = cv2.resize(sr_img, (gt_img.shape[1], gt_img.shape[0]))

    psnr = calculate_psnr(sr_img, gt_img, crop_border=4)
    ssim = calculate_ssim(sr_img, gt_img, crop_border=4)

    print(fname, f"PSNR: {psnr:.2f}, SSIM: {ssim:.4f}")

    psnr_list.append(psnr)
    ssim_list.append(ssim)
    
np.mean(ssim_list), np.mean(psnr_list)
```
